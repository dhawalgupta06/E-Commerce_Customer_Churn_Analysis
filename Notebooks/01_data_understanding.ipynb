## 1. Data Understanding

The objective of this section is to understand the structure, variables, and overall quality of the e-commerce customer dataset before performing any cleaning or analysis.


### Import Required Library

Pandas is imported for loading, inspecting, and analyzing the dataset.


```python
import pandas as pd
```

### Load the Dataset

The customer-level data is loaded from the `E Comm` worksheet of the original Excel workbook.


```python
df = pd.read_excel('../Data/E Commerce Dataset.xlsx', sheet_name="E Comm")
```

### Dataset Preview

The first few records are reviewed to confirm that the dataset was loaded correctly and to obtain an initial view of its structure and variables.


```python
df.head()
```

### Dataset Structure

The dataset dimensions and column names are inspected to understand the number of customer records and variables available for analysis.


```python
df.shape
```

The dataset contains **5,630 customer records and 20 features** describing customer demographics, engagement, purchasing behavior, service experience, and churn status.


```python
df.columns
```

### Data Types and Completeness

The dataset structure is examined to identify the data type of each variable and obtain an initial view of data completeness.

Columns with fewer than 5,630 non-null observations indicate the presence of missing values that require further assessment.


```python
df.info()
```

### Missing Value Assessment

Missing values are quantified by both count and percentage to understand the extent of incomplete data across the dataset.


```python
df.isnull().sum()
```

```python
((df.isnull().sum() / len(df)) * 100).round(2)
```

Missing values are present in seven numerical columns:

- `Tenure`
- `WarehouseToHome`
- `HourSpendOnApp`
- `OrderAmountHikeFromlastYear`
- `CouponUsed`
- `OrderCount`
- `DaySinceLastOrder`

Each affected column contains approximately **4.5%–5.5% missing observations**. No treatment is performed at this stage; the appropriate approach will be determined during data cleaning.


### Duplicate Assessment

The dataset is checked for both completely duplicated records and duplicated `CustomerID` values.

Because `CustomerID` is expected to uniquely identify each customer, checking its uniqueness helps verify that individual customers are not represented multiple times.


```python
df.duplicated().sum()
```

```python
df['CustomerID'].duplicated().sum()
```

No fully duplicated records or duplicated CustomerIDs were identified. Therefore, each of the **5,630 records represents a unique customer**.


### Numerical Data Assessment

Descriptive statistics are reviewed primarily as a data-quality check to examine the ranges and general characteristics of the numerical variables.

Minimum and maximum values are inspected to identify potentially unusual or implausible observations that may require further investigation.


```python
df.describe()
```

### Investigation of Unusually High Values

Several numerical variables show relatively high maximum values compared with their typical ranges. Their highest observations are reviewed before deciding whether they represent data-quality issues.

Extreme observations are not automatically treated as errors simply because they are unusually high.


```python
columns = [
    'Tenure',
    'WarehouseToHome',
    'NumberOfAddress',
    'CouponUsed',
    'DaySinceLastOrder',
    'OrderCount'
]

for col in columns:
    print(f"\n{col}")
    print(df[col].sort_values(ascending=False).head(10).values)
```

The upper-range observations were inspected and there was insufficient evidence to classify them as invalid. These values will therefore be retained rather than removed solely because they are unusually high.


### Categorical Data Assessment

Categorical variables are inspected to identify inconsistent labels, duplicate representations of the same category, and potentially invalid values.


```python
categoricalColumns = [
    'PreferredLoginDevice',
    'PreferredPaymentMode',
    'Gender',
    'PreferedOrderCat',
    'MaritalStatus'
]

for col in categoricalColumns:
    print(f"\n{col}")
    print(df[col].unique())
```

Three categorical variables contain inconsistent labels representing the same underlying categories:

- `PreferredLoginDevice`: `Phone` and `Mobile Phone`
- `PreferedOrderCat`: `Mobile` and `Mobile Phone`
- `PreferredPaymentMode`: `CC` and `Credit Card`; `COD` and `Cash on Delivery`

These inconsistencies will be standardized during the data-cleaning stage.


## Data Quality Findings

The initial data-understanding process identified the following findings:

1. The dataset contains **5,630 customer records and 20 variables**.
2. Each record represents a unique customer, with **no duplicate rows or duplicate CustomerIDs**.
3. **Seven numerical columns contain missing values**, with approximately 4.5%–5.5% missing observations in each affected column.
4. Numerical ranges were reviewed for potentially unusual observations. The investigated upper-range values were retained because there was insufficient evidence to classify them as invalid.
5. **Three categorical columns contain inconsistent category labels** that require standardization:
   - `PreferredLoginDevice`
   - `PreferedOrderCat`
   - `PreferredPaymentMode`
6. No transformations or cleaning were performed during this stage. The identified data-quality issues will be addressed separately during the data-cleaning process.
