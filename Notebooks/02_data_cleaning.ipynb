# E-Commerce Customer Churn & Retention Analysis

## 2. Data Cleaning

The objective of this stage is to address the data-quality issues identified during data understanding and prepare a consistent, complete dataset for analysis.

The cleaning process focuses on two areas:

- standardizing inconsistent categorical labels;
- treating missing values in the affected numerical variables.

The cleaned dataset is then validated to confirm that the identified issues have been resolved without removing customer records.


### Import Required Library

Pandas is used to load and clean the customer dataset.


```python
import pandas as pd
```

### Load the Raw Dataset

The original `E Comm` worksheet is loaded again so that the cleaning process begins from the raw customer data and remains separate from the data-understanding notebook.


```python
df = pd.read_excel('../Data/E Commerce Dataset Raw.xlsx', sheet_name="E Comm")
```

```python
df.head()
```

## Categorical Label Standardization

The data-understanding stage identified multiple labels representing the same underlying categories. These values are standardized to create consistent categorical variables before analysis.


### Preferred Login Device

`Mobile Phone` is standardized to `Phone` so that both labels are represented by a single category.


```python
df["PreferredLoginDevice"] = df["PreferredLoginDevice"].replace("Mobile Phone", "Phone")
```

### Preferred Payment Mode

Abbreviated payment labels are replaced with their full descriptions:

- `CC` → `Credit Card`
- `COD` → `Cash On Delivery`

This prevents the same payment method from being treated as separate categories during analysis.


```python
df["PreferredPaymentMode"] = df["PreferredPaymentMode"].replace("CC", "Credit Card")
df["PreferredPaymentMode"] = df["PreferredPaymentMode"].replace("COD", "Cash On Delivery")
```

### Preferred Order Category

`Mobile Phone` is standardized to `Mobile`, leaving a single consistent label for the mobile product category.


```python
df["PreferedOrderCat"] =df["PreferedOrderCat"].replace("Mobile Phone", "Mobile")
```

## Missing Value Treatment

Seven numerical variables contained missing observations.

Median imputation is used for `Tenure`, `WarehouseToHome`, `HourSpendOnApp`, `CouponUsed`, `OrderCount`, and `DaySinceLastOrder`. This provides a representative central value while limiting the influence of unusually high observations and, where relevant, preserving the interpretation of discrete variables.

`OrderAmountHikeFromlastYear` is imputed using its mean because the variable represents a percentage measure and the average provides an appropriate central estimate for the missing observations.


```python
df["Tenure"]= df["Tenure"].fillna(df["Tenure"].median())
df["WarehouseToHome"] = df["WarehouseToHome"].fillna(df["WarehouseToHome"].median())
df["HourSpendOnApp"]= df["HourSpendOnApp"].fillna(df["HourSpendOnApp"].median())
df["CouponUsed"] = df["CouponUsed"].fillna(df["CouponUsed"].median())
df['OrderCount']=df["OrderCount"].fillna(df["OrderCount"].median())
df["DaySinceLastOrder"]=df["DaySinceLastOrder"].fillna(df["DaySinceLastOrder"].median())
df["OrderAmountHikeFromlastYear"]=df["OrderAmountHikeFromlastYear"].fillna(df["OrderAmountHikeFromlastYear"].mean())
```

## Cleaning Validation

After applying the transformations, the cleaned dataset is validated to confirm that categorical inconsistencies and missing values have been resolved and that record integrity has been preserved.


### Validate Standardized Categories

The unique values in each cleaned categorical column are reviewed to confirm that the duplicate category labels no longer exist.

```python
df["PreferredPaymentMode"].unique()
```

```python
df["PreferedOrderCat"].unique()
```

The standardized categorical variables now contain one consistent label for each underlying category.


### Validate Missing Values

All columns are checked again for null values to confirm that the missing-value treatment was successful.


```python
df.isnull().sum()
```

No missing values remain in the cleaned dataset.


### Validate Record Integrity

`CustomerID` uniqueness and full-row duplicates are checked again to ensure that the cleaning process did not introduce duplicate customer records.


```python
df["CustomerID"].duplicated().sum()
```

```python
df.duplicated().sum()
```

No duplicated CustomerIDs or fully duplicated rows are present after cleaning.


### Validate Dataset Completeness

The final non-null record count is reviewed across all columns to confirm that the dataset still contains all 5,630 customer records after cleaning.


```python
df.count()
```

## Data Cleaning Summary

The data-cleaning stage successfully prepared the dataset for analysis:

1. Inconsistent labels were standardized in `PreferredLoginDevice`, `PreferredPaymentMode`, and `PreferedOrderCat`.
2. Missing values in seven numerical variables were treated using mean or median imputation based on the characteristics of each variable.
3. No missing values remain.
4. No duplicate rows or duplicate CustomerIDs are present.
5. All **5,630 customer records** were retained.

The dataset is now clean, consistent, and ready for exploratory analysis of customer churn and retention.
